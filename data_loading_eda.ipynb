SUMMARY

Target imbalance: The dataset shows significant class imbalance — only 16.1% of employees have Attrition = Yes. This will need to be addressed (e.g., via SMOTE) before model training, and accuracy alone will be a misleading metric for evaluation.

Columns to drop: EmployeeCount, StandardHours, and Over18 are zero-variance (single unique value across all rows). EmployeeNumber is a unique identifier with no predictive value. All four will be dropped in preprocessing.

Strongest categorical predictors observed: OverTime shows the sharpest split — employees working overtime leave at ~30.5% vs. 10.4% for those who don't (~3x higher). JobRole also shows wide variation, with Sales Representatives at ~39.8% attrition vs. Research Directors at ~2.5%. MaritalStatus and BusinessTravel both show clear gradients (Single > Married > Divorced; Travel_Frequently > Travel_Rarely > Non-Travel).

Skew observations: MonthlyIncome and YearsAtCompany are both strongly right-skewed, with most employees clustered at lower values and long tails toward higher values. DailyRate is unusually close to uniformly distributed. Age is roughly normal with slight right skew. These skew patterns may affect Logistic Regression, KNN, and Naive Bayes more than tree-based models.

Multicollinearity flagged: Two clusters of highly correlated features (>0.7) were identified — (1) JobLevel, MonthlyIncome, TotalWorkingYears (JobLevel–MonthlyIncome at 0.95 is especially high), and (2) YearsAtCompany, YearsInCurrentRole, YearsWithCurrManager. All features will be retained for now given tree-based models in the comparison are robust to multicollinearity, but this will be revisited if Logistic Regression coefficients appear unstable.

Ordinal features: JobSatisfaction, WorkLifeBalance, EnvironmentSatisfaction, RelationshipSatisfaction (1–4 scales), Education (1–5 scale), and PerformanceRating are already numerically encoded and require no additional mapping. Notably, PerformanceRating only contains values 3 and 4 in practice, meaning it may carry limited predictive signal.
